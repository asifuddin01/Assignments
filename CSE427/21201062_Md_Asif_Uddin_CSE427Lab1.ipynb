```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Create two linear arrays a and b with 20 random elements each ranging from 1 to 100.
a = np.random.randint(1, 101, 20)
b = np.random.randint(1, 101, 20)
print("Array a:", a)
print("Array b:", b)
```

```python
# 2. change those linear arrays to two 5x4 matrices.
a = a.reshape(5, 4)
b = b.reshape(5, 4)

print("\nMatrix a:\n", a)
print("Matrix b:\n", b)
```

```python
# 3. Perform scalar multiplication with matrix a and b.
print("\nScalar Multiplication of a by 2:\n", a * 2)
print("Scalar Multiplication of b by 3:\n", b * 3)
```

```python
# 4. modify the matrix b and perform matrix multiplication with a and b to store it in c.
b[0, 0] = 1  # modify one value just to show change
c = np.dot(a, b.T)  # matrix multiplication (a × bᵀ)
print("\nMatrix c (a × bᵀ):\n", c)
```

```python
# 5. Print only column 2 and 3 of matrix c
print("\nColumns 2 & 3 of c:\n", c[:, 1:3])
```

```python
# Print the whole matrix using slicing

print(c[:, :])
```

```python
# 6. Find the largest element in the array, and also the index of that element.
max_value = c.max()
max_index = np.unravel_index(np.argmax(c), c.shape)

# Convert to plain Python int
row, col = int(max_index[0]), int(max_index[1])

print("\nLargest element in c:", max_value)
print(f"Index of largest element: ({row}, {col})")
```

```python
# 7. Convert c into 1D array
c_flat = c.flatten()
print("\nOne-dimensional version of c:\n", c_flat)
```

```python
# Data
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28,25,29,20,26,50,19,29,52,55]

study_hours = [2,3,4,4,5,6,7,7,8,9,9,10,11,11,12]
marks = [6,10,15,20,34,44,55,60,55,67,70,80,90,99,100]

subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
subject_marks = [89, 90, 45, 78, 99]
```

```python
# 1. Scatter plot
plt.figure(figsize=(5, 3))
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.title("Temperature vs Humidity (Dhaka)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.show()
```

```python
# 2. Line plot

plt.figure(figsize=(5, 3))
plt.plot(study_hours, marks, marker='o', color='blue')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#3. Horizontal Bar Graph
plt.figure(figsize=(5, 3))
plt.barh(subjects, subject_marks, color='orange')
plt.title("Horizontal Bar Graph - Subject Scores")
plt.xlabel("Marks")
plt.ylabel("Subjects")
plt.show()
```

```python
# 4. Vertical Bar Graph
plt.figure(figsize=(5, 3))
plt.bar(subjects, subject_marks, color=['red', 'green', 'blue', 'purple', 'cyan'])
plt.title("Vertical Bar Graph - Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
# All Graphs Together Using subplot()

plt.figure(figsize=(10, 8))

# 1
plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.title("Temperature vs Humidity (Dhaka)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")

# 2
plt.subplot(2, 2, 2)
plt.plot(study_hours, marks, marker='o', color='blue')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# 3
plt.subplot(2, 2, 3)
plt.barh(subjects, subject_marks, color='orange')
plt.title("Horizontal Bar Graph")
plt.xlabel("Marks")
plt.ylabel("Subjects")

# 4
plt.subplot(2, 2, 4)
plt.bar(subjects, subject_marks, color=['red','green','blue','purple','cyan'])
plt.title("Vertical Bar Graph")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.tight_layout()
plt.show()
```
